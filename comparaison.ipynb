```python
from Bio import SeqIO
import pandas as pd

# Charger les fichiers GenBank
file_ce = "./data/S_cerevisiae.gb"
file_sp = "./data/S_pombe.gb"

record_ce = SeqIO.read(file_ce, "genbank")
record_sp = SeqIO.read(file_sp, "genbank")
```

```python
from Bio.SeqRecord import SeqRecord

def extract_cds_annotations(genbank_file):
    record = SeqIO.read(genbank_file, "genbank")

    cds_records = []
    annotations = []

    for feature in record.features:
        if feature.type == "CDS":
            sequence = feature.extract(record.seq)
            
            annot = {key : value[0] for key, value in feature.qualifiers.items()}
            annot["location"] = feature.location
            annot["seq_len"] = len(sequence)
            
            cds_records.append(SeqRecord(sequence, name=annot["gene"], description=annot["product"]))
            annotations.append(annot)

    return cds_records, annotations
```

```python
r_ce, a_ce = extract_cds_annotations(file_ce)
r_sp, a_sp = extract_cds_annotations(file_sp)
```
